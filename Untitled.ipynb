# Galaxies & Their Environments Code
#### Calculating $F_\nu = \frac{F_\lambda \lambda^2}{c}$

```python
import astropy.units as u
from astropy import constants as const
import numpy as np
```

```python
g_avg = 4.4E-12 / u.cm**2 * u.erg / u.second / u.Angstrom
u_avg = 1.5E-12 / u.cm**2 * u.erg / u.second / u.Angstrom 
g_len = 4825 * u.Angstrom
u_len = 3557 * u.Angstrom

F_gnu = g_avg * g_len**2 / const.c.to(u.Angstrom/u.s)
F_unu = u_avg * u_len**2 / const.c.to(u.Angstrom/u.s)

print(F_gnu)
print(F_unu)
```

```python
Where erg / cm2 is equivalent to erg / cm2 / Hz / s
```

Calculating Colour
### $m_{bluer} - m_{redder} = 2.5\cdot\log10{\frac{filterflux_{red}}{filterflux_{blue}}}$

```python
delta_m = 2.5*np.log10(F_gnu/F_unu)
print(delta_m)
```
